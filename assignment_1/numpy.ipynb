Part 4 – NumPy

```python
import numpy as np
```

1) Create an array containing the numbers from 1 to 10.

```python
numbers = np.arange(1, 11)
```

2) Print the array.

```python
print(numbers)
```

#3) Print its minimum and maximum value.


```python
print(numbers.min())
print(numbers.max())
```

#4) Create an array of 5 zeros and print it.


```python
zeros=np.zeros(5)
print(zeros)
```

#5) Create an array of 5 ones and print it.


```python
ones=np.ones(5)
print(ones)
```

#6) Create an specific 2x3 matrix and print it.


```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

#7) Display the matrix.


```python
print(matrix)
```

#8) Use .shape to display its dimensions.


```python
print(matrix.shape)
```
