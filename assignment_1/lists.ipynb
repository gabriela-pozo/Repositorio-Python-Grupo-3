Part1-Lists

1. Print the original list

```python
grades = [15, 18, 12, 20, 16]
print("Original list:", grades)
```

2. Add the grade 17 ussing(append).

```python
grades.append(17)
print(grades)
```

3. Sort the grades from lowest to highest using sort()

```python
grades.sort()
print(grades)
```

4. Display information of grades

```python
print("The highest grade:", max(grades))
print("The lowest grade:", min(grades))
print("The total number of grades:", len(grades))
```

5. Print the final list

```python
print("The final list", grades)
```
