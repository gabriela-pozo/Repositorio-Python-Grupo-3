### Part 2 - Tuples

0. Create a tuple

```python
ages=(20, 25, 22, 30, 28)
```

1. Print the tuple

```python
print(ages)
```

2. Display the first and last elements

```python
# The first element
print("The first element is",ages[0])


# The last element
print("The last element is",ages[-1])
```

3. Display:
- the maximum value
- the minimum value
- the number of elements

```python
# The maximum value
print("The maximum value is",max(ages))

# The minimum value
print("The minimun value is",min(ages))

# The number of elements
print("The number of elements is",len(ages))
```
